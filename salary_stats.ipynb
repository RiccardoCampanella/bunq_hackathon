```python
import json
from collections import defaultdict

# Load the transactions data
file_path = "/Users/riccardocampanella/Documents/hackathon_data/synthetic_transactions.json"
with open(file_path) as f:
    transactions = json.load(f)

# Filter transactions with category 'SALARY'
salary_transactions = [txn for txn in transactions if txn.get('category') == 'SALARY']

# Group transactions by owner_user_id
grouped_by_user_id = defaultdict(list)
for txn in salary_transactions:
    user_id = txn.get('owner_user_id')
    if user_id:
        grouped_by_user_id[user_id].append(txn)

# Save grouped transactions to a new JSON file
output_file_path = "C:\\Users\\aryan\\Downloads\\hackathon_2024\\hackathon_data\\grouped_salary_transactions.json"
with open(output_file_path, "w") as f:
    json.dump(grouped_by_user_id, f, indent=2)

print(f"Grouped transactions saved to {output_file_path}")
```

```python
import json
from datetime import datetime

file_path = "C:\\Users\\aryan\\Downloads\\hackathon_2024\\hackathon_data\\grouped_salary_transactions.json"
import json
import csv

# Load JSON data
with open('C:\\Users\\aryan\\Downloads\\hackathon_2024\\hackathon_data\\grouped_salary_transactions.json', 'r') as f:
    data = json.load(f)

# Extract relevant information
extracted_data = []
for entry_id, transactions in data.items():
    for transaction in transactions:
        if transaction.get('category') == 'SALARY':
            extracted_data.append({
                'userID': transaction['owner_user_id'],
                'salary': transaction['amount'],
                'timestamp': transaction['updated_timestamp']
            })

# Save to CSV
with open('salary_data.csv', 'w', newline='') as csvfile:
    fieldnames = ['userID', 'salary', 'timestamp']
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    
    writer.writeheader()
    writer.writerows(extracted_data)

print(f"Extracted {len(extracted_data)} salary entries to salary_data.csv")
```
